1. Count how many vowels are in a given string.

```python
user_input = input('Enter the string!: ')
vowels = ('a','e','i','o','u')
count = 0

for ch in user_input:
  if ch.lower() in vowels:
    count += 1

print(f'Vowel count in the string {user_input}: {count}')
```

2. Reverse a number using only a for loop (no slicing).

Example: Input → 12345, Output → 54321

```python
num = 12345
rev = 0

for i in range(len(str(num))):
  digit = num % 10
  rev = rev * 10 + digit
  num = num // 10

print(rev)
```

3. Find the sum of digits of a number.

Example: 5483 → 20

```python
num = 5483
sum_num = 0
for ch in str(num):
  sum_num += int(ch)

print(sum_num)
```

5. Find the factorial of a number.

```python
num = int(input('Enter the number for factorial : '))
fact = 1
for i in range(1,num+1):
  #print(i)
  fact *= i

print(f'Factorial of {num} : {fact}')
```

6. Check whether a number is prime using a for loop.

```python
num = int(input('Enter the number to check if prime or not : '))
if num <= 1:
  print('Not prime!')
else:
  is_prime = True
  for i in range(2, int(num ** 0.5) + 1):
    if num % i == 0:
      is_prime = False
      break

  if is_prime:
    print('Prime Number!')
  else:
    print('Not a Prime Number')
```

9. Find the largest number in a list.

Example: [4, 7, 1, 9, 3]

```python
arr = [4,7,1,9,3]
max_num = arr[0]

for num in arr:
  if num > max_num:
    max_num = num

print(max_num)
```

11. From 1 to 50, find:

a) sum of even numbers

b) sum of odd numbers

```python
sum_even = 0
sum_odd = 0

for i in range(1,51):
  if i % 2 == 0:
    sum_even += i
  else:
    sum_odd += i

print(sum_even)
print(sum_odd)

# Way 2 :
sum_even1 = sum(range(2, 51, 2))
sum_odd1 = sum(range(1, 51, 2))

print(sum_even1)
print(sum_odd1)
```

13. Check whether a string is a palindrome.
Example: "madam"

```python
s = input('Enter a string to check if it is a palindrome or not : ')
if s == s[::-1]:
  print('Palindrome')
else:
  print('Not Palindrome')
```

14. Find duplicate values in a list.
Example: [1,2,3,2,4,5,1]

```python
nums = [1, 2, 3, 2, 4, 5, 1]
duplicates = []

for num in nums:
  if nums.count(num) > 1 and num not in duplicates:
      duplicates.append(num)

print(duplicates)
```

15. Print the frequency of each character in a string.
Example: "banana"

```python
s = "banana"
checked = ""

for ch in s:
  if ch not in checked:
      print(ch, ":", s.count(ch))
      checked += ch
```

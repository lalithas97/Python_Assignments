### Problem 3.5: Armstrong Number (Hard)

Write a program to check if a number is an Armstrong number (also called Narcissistic number).

An Armstrong number is a number that is equal to the sum of its own digits each raised to the power of the number of digits.

Examples:
- 153 = 1³ + 5³ + 3³ = 1 + 125 + 27 = 153 (Armstrong number)
- 9474 = 9⁴ + 4⁴ + 7⁴ + 4⁴ = 6561 + 256 + 2401 + 256 = 9474 (Armstrong number)

**Hint:** First count digits (power). Store original. Extract each digit, raise to power, add to sum. Compare sum with original.

````python
# Your code here
num = int(input('Enter a number to check whether it is an armstrong number or not : '))
digits = str(num)
power = len(digits)
total = 0

for d in digits:
    total += int(d) ** power

if total == num:
    print(f'{num} is an Armstrong Number')
else:
    print(f'{num} is not an Armstrong Number')
````

### Problem 4.6: Diamond Pattern (Hard)

Print a diamond pattern for n rows (n should be odd):

```
    *
   ***
  *****
 *******
*********
 *******
  *****
   ***
    *
```
(Example shown for n=9)

**Hint:** Print pyramid (upper half) for rows 1 to (n+1)//2, then print inverted pyramid (lower half) for remaining rows.

````python
# Your code here (R)
n = int(input('Enter an odd num : '))

if n%2 == 0:
    print('Please enter the odd number')
else:
    middle = (n + 1) // 2

    #Upper half
    for row in range(1, middle+1):
        spaces = middle - row
        stars = 2 * row - 1
        print(" " * spaces + "*" * stars)
    
    #Lower Half
    for row in range(middle + 1, n + 1):
        spaces = row - middle
        stars = 2 * (n - row) + 1
        print(" " * spaces + "*" * stars)
````

### Problem 6.2: GCD Calculator (Medium)

Write a program to find the Greatest Common Divisor (GCD) of two numbers using the Euclidean algorithm.

Algorithm:
1. Divide the larger number by the smaller number
2. Replace the larger number with the smaller number
3. Replace the smaller number with the remainder
4. Repeat until the remainder is 0
5. The GCD is the last non-zero remainder

**Hint:** While b != 0: store remainder (a%b), then a=b, b=remainder. When loop ends, a is the GCD.

````python
# Your code here(R)
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

while b != 0:
    remainder = a % b
    a = b
    b = remainder

print("GCD is:", a)
````

### Problem 6.6: Strong Number Checker (Hard)

Write a program to check if a number is a Strong number.

A Strong number is a number whose sum of factorial of digits is equal to the original number.

Example: 145 is a strong number because 1! + 4! + 5! = 1 + 24 + 120 = 145

**Hint:** Extract each digit, calculate its factorial, add to sum. Compare sum with original number.

````python
# Your code here
n = int(input('Enter a number if it is a strong number or not : '))
sum_digits = 0
temp = n
while n > 0:
    digit = n%10

    #Factorial of digit
    fact = 1
    for i in range(1, digit+1):
        fact *= i
    
    sum_digits += fact

    n //= 10

if temp == sum_digits :
    print('Strong Number!')
else:
    print('Not Strong Number!')
````

The Caesar cipher shifts each letter by a fixed number of positions in the alphabet.

Example: With shift=3
- 'A' becomes 'D'
- 'Hello' becomes 'Khoor'

Your program should:
1. Ask for a message and shift value
2. Provide options to encrypt or decrypt
3. Handle both uppercase and lowercase letters
4. Keep non-alphabetic characters unchanged

### Problem 7.3: ATM Machine Simulator (Hard)

Create an ATM machine simulator with the following features:
1. Initial balance: 10000
2. PIN: 1234 (3 attempts allowed)
3. Menu options:
   - Check Balance
   - Withdraw Money (check for sufficient balance and valid amount)
   - Deposit Money
   - Change PIN
   - Exit
4. Keep transaction history (last 5 transactions)
5. Input validation for all operations

**Hint:** Use variables for balance, pin. Loop for PIN attempts. Use while loop for menu. Use list to store transaction history. Validate amounts > 0 and balance >= withdrawal.

````python
# Your code here
'''
Step 1 : This program keeps on running
Step 2 : Verification
    - PIN is verified
    - What if PIN is wrong?
        - Limit exceeds (3 times)
        - if exceeded EXIT
    A loop for PIN attempts. A counter. A condition to block access.
Step 3 : After PIN is correct, ATM should show menu
    - 1. Check Balance: 
        - just print the amount present
    - 2. Withdraw Money 
        - Withdrawal amount should be positive
        - Withdrawal amount can't exceed the balance
        - Decrease the balance
    - 3. Deposit Money:
        - Can't deposit zero or negative
        - Increase the balance
    - 4. Change PIN
        - Old PIN should be verified.
        - New PIN should be stored (reassign the value)
    - 5. Exit
        - Break the loop
Step 4 : Transaction History
    - Store only 5
    - What do we add to history?
        - Withdraw
        - Deposit
    So Just append the transactions
    If the length > 5 ---> Remove the first
'''
````

````python
#Initial Values
balance = 10000
pin = 1234
attempts = 3
transaction_history = []

# ---------- PIN VERIFICATION ----------------
while attempts > 0:
    entered_pin = int(input('Enter your PIN : '))

    if entered_pin == pin:
        print('PIN Verified Successfully! \n')
        break
    else:
        attempts -= 1
        print(f'Wrong PIN. Attempts left : {attempts}')
    
# If PIN attempts exhausted
if attempts == 0:
    print('Too many wrong attempts. Exiting............')
else:
    # Main Menu
    while True:
        print('\n ---- ATM Menu ----')
        print('1. Check Balance')
        print('2. Withdraw Money')
        print('3. Deposit Money')
        print('4. Change PIN')
        print('5. Transaction History')
        print('6. Exit')

        option = int(input('Choose an option (1 - 6) : '))

        # Option 1 : Check Balance
        if option == 1:
            print(f' Current Balance : ₹{balance}')
        
        # Option 2 : Withdraw Money
        elif option == 2:
            amount = int(input('Enter the amount to withdraw: '))
            
            # Handling edge cases
            if amount < 0:
                print('Invalid amount')
            elif amount > balance:
                print('Insufficient Balance')
            # Withdrawal process
            else:
                balance -= amount
                transaction_history.append(f'Withdrawn ₹{amount}')
                print(f'{amount} withdrawn successfully!')
                
        
        # Option 3: Deposit Money
        elif option == 3:
            amount = int(input('Enter amount to deposit : '))

            if amount < 0 :
                print('Invalid Amount')
            else:
                balance += amount
                transaction_history.append(f' Deposited ₹{amount}')
                print(f'₹{amount} deposited successfully!')
        
        # Option 4: Change PIN
        elif option == 4:
            old_pin = int(input('Enter old PIN : '))
            if old_pin == pin:
                new_pin = int(input('Enter New PIN : '))
                pin = new_pin
                print('PIN Changed Successfully!')
            else:
                print('Incorrect old PIN')
        
        # Option 5 : Transaction History
        elif option == 5:
            print('\n Last Transactions : ')
            if not transaction_history:
                print(' No transactions yet.')
            else:
                for transaction in transaction_history[-5:]:
                    print(transaction)
        
        # Option 6: EXIT
        elif option == 6:
            print(' Exiting! Thank you!')
            break

        else:
            print('Invalid Option. Please choose between 1-6')

        
        #Keep only last 5 transactions
        if len(transaction_history) > 5:
            transaction_history.pop(0)
````

A magic number is one where repeatedly summing digits eventually gives 1.
Example: 19 → 1+9=10 → 1+0=1

**Hint:** While n > 9: calculate sum of digits, n = sum. After loop, check if n == 1.

````python
# Your code here
n = int(input('Enter a number : '))

#Keep reducing the number untill it becomes a single digit
while n > 9:
    temp = n 
    digit_sum = 0 

    #Calculate sum of digits of the number
    while temp > 0:
        digit_sum += temp % 10 #Add last digit
        temp //= 10  # Remove last digit

    #Update n with the sum of digits
    n = digit_sum

if n == 1:
    print('Magic Number')
else:
    print('Not a Magic Number')
````

### Problem 10.29: ISBN Validator (Hard)

Validate a 10-digit ISBN number using the check digit algorithm.

**Hint:** ISBN-10: sum = (d1*10 + d2*9 + d3*8 + ... + d10*1). Valid if sum % 11 == 0. Loop with position weight.

````python
# Your code here
'''
1. Read ISBN as a string
2. Loop through each character
3. Multiply digit by its position (1 to 10)
4. Treat 'X' as 10 only if it is the last digit
5. Check if total sum is divisible by 11
'''

isbn = input('Enter ISBN-10 number: ')

#ISBN must be exactly 10 characters
if len(isbn)!= 10:
    print('Invalid ISBN')
else:
    total = 0

    for i in range(10):
        char = isbn[i]

        #If last character is 'X', treat it as 10
        if char == 'X' and i == 9:
            value = 10
        
        # If character is a digit
        elif char.isdigit():
            value = int(char)
        else:
            print('Invalid ISBN')
            break

        # Multiply digit with its position(i + 1)
        total += value * (i+1)

    #Check ISBN validity
    if total % 11 == 0:
        print('Valid ISBN')
    else:
        print('Invalid ISBN')
````

### Problem 10.30: Kaprekar Number (Hard)

A Kaprekar number is one where square can be split into parts that sum to the original number.
Example: 45 (45² = 2025, 20+25 = 45)

**Hint:** Calculate square. Split into right part (last n digits) and left part (remaining). Check if left + right == original.

````python
# Your code here(R)
'''
A number is a Kaprekar number if :
- You square the number
- Split the square into two parts
- The sum of the parts equals the original number
'''

n = int(input('Enter a number : '))

#Square the number
square = n * n

#Count number of digits in n
temp = n
digits = 0
while temp > 0:
    digits += 1
    temp //= 10

#Calculate 10^digits
power = 1
for _ in range(digits):
    power *=  10

#Split the square
right = square % power #Last 'digits' digits
left = square // power #Remaining digits

#Check Kaprekar condition
if left + right == n:
    print('Kaprekar Number')
else:
    print('Not a Kaprekar Number')
````
